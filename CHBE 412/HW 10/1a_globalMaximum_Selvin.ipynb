Find the global maximum of the function $f$

$$
f(x) = g(x-25.4)+1840 \sin(2x)
$$

where

$$
g(x) = -x^4 + 63.4 x^3 - 1277.15 x^2 + 8618.28 x - 15463.1
$$

```python
import numpy as np
import scipy.optimize as opt
import matplotlib.pyplot as plt


def g(x):
    return -x**4 + 63.4*x**3 - 1277.15*x**2 + 8618.28*x - 15463.1 

def f(x):
    return g(x-25.4) + 1840*np.sin(2*x)

lower = 20
upper = 60


soln = opt.minimize(lambda x: -f(x),50, bounds = [[lower,upper]]) #dependent on the initial guess given
gmax = soln.x[0]

x = np.linspace(-10,100,10000)
y = np.array([f(xi) for xi in x])

plt.plot(x,y,'r-')
plt.plot(gmax,f(gmax),'bo')
print('The global maxium of the function: %4.3f' % f(gmax))
plt.plot()
plt.xlim(20,65)
plt.ylim(-13000,5000)
```
