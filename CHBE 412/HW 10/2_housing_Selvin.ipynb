A housing developer owns 50 lots in a new subdivision. They can build two townhomes or one house on each lot. Townhomes have an average cost of \\$120k while houses cost \\$200k, on average, to build. The median sale price of the homes will be \\$400k. 

The median sale price of the townhomes depends on the fraction of the lots that hold townhomes. If there are only two townhomes, it is expected they will sell for \\$250k each. If the entire subdivision is townhomes, it is expected they will sell for \\$150k each. Assume the relationship is linear.

Use the function `minimize()` to determine how many townhomes and houses the developer should build to maximize their profit. Report the expected townhome price and overall profit for this scenario.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as opt

def TH_Price(units):
    return -(100/98)*units + 252.04

def profit(H):
    townhomes,house = H 
    house_profit = house*400
    townhomes_profit = townhomes*TH_Price(townhomes)
    cost = townhomes*120 + house*200
    return townhomes_profit + house_profit - cost

def subdivision(H):
    townhomes,house = H 
    return 100 - townhomes - 2*house

bnds = [[0,None]]*2
cons = [{'type':'ineq','fun': subdivision}]
sol = opt.minimize(lambda H: -profit(H), [1,1], bounds = bnds, constraints = cons)

townhomes,house = sol.x

print('Townhomes:',round(townhomes))
print('Houses:',round(house))
print(f'Townhome Price: ${TH_Price(townhomes):.2f}k')
print(f'Profit: ${profit(sol.x):.2f}k')
```
