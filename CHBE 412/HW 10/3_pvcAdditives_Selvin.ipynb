A company produces 20 tonnes of PVC at a cost of $0.65/kg, from which it manufactures two types of piping by mixing three different additives.

The additive data is:

| Additive | Inventory (kg) | Cost ($/kg) |
| :---: | :--: | :---: |
| 1     | 5000 | 0.80 |
| 2     | 1000 | 0.96 |
| 3     | 2000 | 1.03 |


The product data is:

|Product | Mass Specification | Price ($/kg) |
| --- | :--- | --- |
| A | Not less than 1% of 1 | 1.05 |
|   | Not more than 10% of 2 + 3 | |
| B | Not less than 35% of 1 | 1.25 |
|   | Not less than 42% of 1 + 3 | |

Report the amount and composition of each product that will maximize the profit.

```python
import numpy as np
import scipy.optimize as opt

def profit(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    revenue = 1.05*(A1+A2+A3+A_pvc) + 1.25*(B1+B2+B3+B_pvc)
    cost = 0.80*(A1+B1) + 0.96*(A2+B2) + 1.03*(A3+B3)
    cost_PVC = 0.65*(A_pvc+B_pvc)
    return revenue -cost -cost_PVC

def one(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return 5000 - A1 - B1

def two(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return 1000 - A2 - B2

def three(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return 2000 - A3 - B3

def productA_1(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return A1 - 0.01*(A1+A2+A3+A_pvc)   # A1 can not be less than 1% of total A

def productA_23(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return 0.10*(A1+A2+A3+A_pvc) - (A2+A3)  # A2 and A3 can not be more than 10% of total A


def productB_1(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return B1- 0.35*(B1+B2+B3+B_pvc)   # B1 can not be less than 35# of total B

def productB_13(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return (B1+B3) - 0.42*(B1+B2+B3+B_pvc) # B1 and B3 can not be less than 42% of total B

def PVC_avail(v):
    A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = v
    return 20000 - (A_pvc+B_pvc)


# main code

bnds = [[0,None]]*8

cons = [{'type':'ineq','fun':one},
        {'type':'ineq','fun':two},
        {'type':'ineq','fun':three},
        {'type':'ineq','fun':productA_1},
        {'type':'ineq','fun':productA_23},
        {'type':'ineq','fun':productB_1},
        {'type':'ineq','fun':productB_13},
        {'type':'ineq','fun':PVC_avail},]

v0 = [1000,1000,1000,1000,1000,1000,10000,10000]
soln = opt.minimize(lambda v: -profit(v),v0,bounds = bnds, constraints = cons)
A1,A2,A3,B1,B2,B3, A_pvc, B_pvc = soln.x
A = A1+A2+A3+A_pvc
B = B1+B2+B3+B_pvc

print(f'Amount A: {A:.2f}kg')
print(f'Composition: A1 = {A1/A:.2f} A2 = {A2/A:.2f} A3 = {A3/A:.2f}')
print()
print(f'Amount B: {B:.2f}kg')
print(f'Composition: B1 = {B1/B:.2f} B2 = {B2/B:.2f} B3 = {B3/B:.2f}')
print()
print(f'Profit = $ {profit(soln.x):.2f}')
```
