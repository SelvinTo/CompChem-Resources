Solve the following ordinary differential equations.

a) 
$$\frac{dx}{dt} = t (x-t)$$

$$\frac{dy}{dt} = \cos(xy)$$

$$x(0) = 1$$

$$y(0) = 1$$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as intx

def myfun(t,v):
    x,y = v   # unpacking the variables into two arrays 
    return t*(x-t), np.cos(x*y)  # unpack them in the order, ex, first equation should be for solving x 

# define time range
ti,tf=0,5
t=np.linspace(ti,tf,1000)

# define initial conditions
v0 = (1,1) # comma at the end to make it a tuple

sol = intx.solve_ivp(myfun,(ti,tf),v0,t_eval=t)   # solve_ivp takes in function, time range, and initial value 

x,y = sol.y
plt.plot(t,y,'b-',lw=3,label='y')
plt.plot(t,x,'r-',lw=3,label='x')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel('t',fontsize=16)
plt.ylabel('x,y',fontsize=16)
plt.legend()
```

b)
$$ y'' + (1-y)y' + y^2 = 0$$

$$y(0) = 10$$

$$y'(0) = 2$$

```python
def res(t,v):
    x1, x2 = v 
    dx1 = x2
    dx2 = -(1-x1)*x2-x1**2
    return dx1,dx2

#timespan
ti,tf=0,10
t = np.linspace(ti,tf,100)

#intial conditions
y0=10
dy0=2
v0 = (y0,dy0)

y,dy = intx.solve_ivp(res,(ti,tf),v0,t_eval=t).y

plt.plot(t,y,'r-',lw=3)
plt.show()
```
