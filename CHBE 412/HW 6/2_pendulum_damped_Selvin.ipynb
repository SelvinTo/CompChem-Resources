A simple pendulum is governed by the differential equation

$$
\frac{d^2\theta}{dt^2} + b \frac{d\theta}{dt} + \frac{g}{l}\text{sin}\theta = 0
$$

where
$\theta$ is the angle of displacement from equilibrium,
$b$ is the damping coefficient,
$g$ is the acceleration due to gravity,
and
$l$ is the length of the pendulum.

The small-angle approximation, valid for $\theta$(0) $\ll$ 1 and $\dot{\theta}$(0) = 0 and b = 0, is

$$
\theta(t) = \theta(0)\ \text{cos}\left(\sqrt{\frac{g}{l}}t\right)
$$

Solve for the motion of a $l$ = 0.9 m pendulum with initial conditions $\theta$(0) = 0.08 and $\dot{\theta}$(0) = 0 and b = 0.4/s.

Compare the numerical solution of the exact ODE to the small-angle approximation.

Sample output is

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as intx

def myfun(t,v):
    b,g,l = 0.4,9.8,0.9
    x1, x2 = v 
    dx1 = x2
    dx2 = -b*x2 -(g/l)*np.sin(x1)
    return dx1,dx2

#timespan
ti,tf=0,10
t = np.linspace(ti,tf,1000)

#intial conditions

y0=0.08
dy0=0
v0 = (y0,dy0)

y,dy = intx.solve_ivp(myfun,(ti,tf),v0,t_eval=t).y


def SAapprx(t):
    O,g,l = 0.08,9.8,0.9
    approx = O*np.cos(np.sqrt(g/l)*t)
    return approx

plt.figure(figsize=(8, 6))
plt.plot(t,y,'r-',lw=3, label = 'b = 0.4/s')
plt.plot(t,SAapprx(t),'b--',lw=3, label = 'approx')
plt.xlabel('Time, $t$ (s)', size = 14)
plt.ylabel('Position, θ', size = 14)
plt.ylim(-0.12,0.12)
plt.legend(loc='lower right')
```
