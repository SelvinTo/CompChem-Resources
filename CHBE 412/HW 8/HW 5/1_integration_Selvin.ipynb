Plot the integrand and evaluate the following integral:

$$I = \int_{-1.2}^{10.2} e^{-t^2/25.6} \sin(t-1.2) dt$$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as intx
```

```python
def f(x):
    return np.exp(-(x**2)/25.6)*np.sin(x-1.2)

x = np.linspace(-3, 15,1000)

I, err = intx.quad(f,-1.2,10.2)
print(f"Integral value {I:.4f} with error {err:.4f}")


plt.plot(x, f(x))
plt.xlabel('x')
plt.ylabel('y')
plt.fill_between(x, f(x), where = [(x>-1.2) and (x<10.2) for x in x], color = 'g', alpha =0.3)
```
