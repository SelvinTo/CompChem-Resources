Find all solutions to the following equations. For each solution, print the initial guess, initial residual(s), the final solution, and final residual(s).

a) $$ x^2 + \sin(4x) = 1$$

```python
import numpy as np
import scipy.optimize as opt

def function1(x):
    F = x**2+np.sin(4*x)-1
    return F

Guess1 = 0.5

z = opt.fsolve(function1,Guess1)
print(f"Initial Guess: {Guess1}, Initial Residual: {function1(Guess1):.5f}, Final Solution: {z[0]:.4f}, Final Residual: {function1(z[0]):.3e}")
```

b) $$ (x-3.1)^2 + y^2 = 12.5 $$
$$ \frac{-x^2}{4} + y = 1 $$

```python
def function2(z):
    x = z[0]
    y = z[1]

    F = np.empty((2))
    F[0] = (x-3.1)**2+y**2-12.5
    F[1] = ((-x**2)/4)+y-1
    return F

Guess2 = np.array([0.5,2])

z = opt.fsolve(function2, Guess2)
finalres = function2(z)
intres = function2(Guess2)

print(f"Initial Guess: X = {Guess2[0]}, Y = {Guess2[1]}, Initial Residual: {intres[0]:.3f},{intres[1]:.3f}, Final Solution: X = {z[0]:.3f}, Y = {z[1]:.3f}, Final Residual: {finalres[0]:.3e}, {finalres[1]:.3e}")
```
