The time it takes for a signal to travel from a GPS satellite to a receiver (e.g. in your smartphone) can be used to triangulate the receiver's position, ($x$, $y$, $z$) in Cartesian coordinates. The GPS signal from satellite $i$ includes the coordinates of the satellite, ($X_i$, $Y_i$, $Z_i$) in Cartesian coordinates, and the satellite's time. The time difference $\Delta t_i$ between the receiver time and the satellite time is the sum of the time it took the signal to travel the distance from satellite to receiver $t^{dist}$ and any error in the receiver time $t^{err}$.

$$
\Delta t_i = t_i^{dist} + t^{err}
$$

The travel time is the distance traveled divided by the velocity of the signal (the speed of light $c$)

$$
\Delta t_i = \frac{\sqrt{(x-X_i)^2 + (y-Y_i)^2 + (z-Z_i)^2}}{c} + t^{err}
$$

There are four unknowns ($x$, $y$, $z$, $t^{err}$), so four statellite signals are needed. Given the four signals in gpsReceiverLocation.csv, find the the four unknowns.

There will be multiple solutions to the set of GPS equations, but only one will be on the surface of the Earth. The receiver's distance from Earth's center is 

$$
r = \sqrt{x^2 + y^2 + z^2}
$$

and the surface is an average of 6371 km from the center. Verify that your solution is near Earth's surface ($\pm$ 100 km).

The location of the receiver can be converted from Cartesian ($x$,$y$,$z$) coordinates to latitude and longitude using the following equations:

$$
\text{Latitude} = \left[ \frac{\pi}{2} - \text{cos}^{-1}\left(\frac{z}{r}\right)\right]\frac{180°}{\pi} \\
\text{Longitude} = \left[ \text{sign}(y)\frac{\pi}{2}-\text{tan}^{-1}\left(\frac{x}{y}\right)\right]\frac{180°}{\pi}
$$

Convert your answer to latitude and longitude. Look up that location online (e.g. using Google Maps). Where is the receiver located?

Hint: km and ms are good internal units for this problem. The numpy functions are `arccos()`, `arctan()`, and `sign()`.

```python
import numpy as np
import os
import scipy.optimize as opt

os.getcwd()

data = np.loadtxt('4_gpsReceiverLocation.csv',delimiter=",", skiprows=3).T

x = data[1]
y = data[2]
z = data[3]
t = data[4]
c = 300 #km/ms


def myfunction(A):
    F = np.empty(4)
    X = A[0]
    Y = A[1]
    Z = A[2]
    T = A[3]
    
    for i in range(len(x)):
        F[i] = -t[i] + np.sqrt((x[i]-X)**2+(y[i]-Y)**2+(z[i]-Z)**2)/c + T
    return F


def check(X,Y,Z):
    r = np.sqrt(X**2+Y**2+Z**2)
    return 6271 < r < 6471

def Coordinates(X,Y,Z):
    r = np.sqrt(X**2+Y**2+Z**2)
    Lat = (np.pi/2-np.arccos(Z/r))*180/np.pi
    Long = (np.sign(Y)*np.pi/2-np.arctan(X/Y))*180/np.pi
    return Lat,Long


while True:
    Guess = np.array([np.random.uniform(-1.9e+3,2.51e+4),np.random.uniform(6.06e+2,2.32e+4),np.random.uniform(-9.13e+3,1e+4),np.random.uniform(7.05e+1,8.08e+1)])
    Coord = opt.fsolve(myfunction, Guess)
    if check(Coord[0],Coord[1],Coord[2]):
       Lat,Long = Coordinates(Coord[0],Coord[1],Coord[2])
       print(f'The Latitude and Longitude coordinates are {Lat:.6f} and {Long:.6f}')
       break

print('The Location is Mt Kilimanjaro, a volcano in Tanzania, Africa')
```
